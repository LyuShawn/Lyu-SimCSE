# 验证数据

```python
import numpy as np
import re

data_str = """
 69.71 & 81.58 & 72.31 & 81.02 & 79.65 & 80.56 & 69.34 
"""

# 使用正则去掉 \textbf{} 标签
cleaned_str = data_str.replace("\textbf{", "").replace("}", "")

# 去除多余的空格和符号，然后将数字转换为浮动数组
arr = np.array([float(i) for i in cleaned_str.replace("&", " ").split()])

# 计算并输出均值，保留两位小数
print(arr.mean().round(2))
```
