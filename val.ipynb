```python
import numpy as np
import re

data_str = """

82.97 88.91 93.21 89.83 87.17 90.92 76.83

"""

# 使用正则去掉 \textbf{} 标签
cleaned_str = data_str.replace("\textbf{", "").replace("}", "")

# 去除多余的空格和符号，然后将数字转换为浮动数组
arr = np.array([float(i) for i in cleaned_str.replace("&", " ").split()])

# 计算并输出均值，保留两位小数
print(arr.mean().round(2))
```

```python
# 计算差值
import numpy as np

data_str = """
            RoBERTa (first-last avg.) & 58.91 & 58.62 & 61.44 & 69.05 & 65.23 & 59.38 & 58.84 & 61.64
            \\
            SimCSE & 72.86 & 83.99 & 75.62 & 84.77 & 81.80 & 81.98 & 71.26 & 78.90
            \\
            ESimCSE & 73.20 & 84.93 & 76.88 & 84.86 & 81.21 & 82.79 & 72.27 & 79.45
            \\
            PromptBERT & 72.89 & \textbf{86.44} & 78.10 & 85.09 & 79.37 & 81.52 & 70.85 & 79.18
            \\
            EASE & 67.83 & 81.62 & 72.90 & 83.45 & 78.27 & 79.17 & 69.80 & 76.15
            \\
            ConPVP & \textbf{74.75} & 84.09 & 77.88 & 83.13 & \textbf{83.44} & 83.64 & \textbf{74.31} & 80.18
            \\
            Out method & 73.09 & 85.74 & \textbf{78.92} & \textbf{86.22} & 82.10 & \textbf{83.82} & 74.20 & \textbf{80.58}
"""
# 按行分割
data_str = data_str.split("\n")
data_list = []
for line in data_str:
    if '&' not in line:
        continue
    cleaned_str = line.replace("\textbf{", "").replace("}", "")
    data = float(cleaned_str.split("&")[-1])
    data_list.append(data)
# 计算最后一个数据和其他数据的差值
data_list = np.array(data_list)
diff = data_list[-1] - data_list[:-1]
print(diff)
```
